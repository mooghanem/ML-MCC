# House Sales Data Analysis Task

## Section 1: Pandas (Cells 1 - 5)

### Cell 1: Library Imports & Initial Data Inspection
**Requirements:**
1. Import `pandas`, `numpy`, and `matplotlib.pyplot`.
2. Load `kc_house_data.csv` into a DataFrame named `df`.
3. Display the first 10 rows and the last 5 rows.
4. Print the dataset shape, column data types, and the total count of missing values per column.

**Tools & Functions to Use:** `pd.read_csv()`, `head()`, `tail()`, `shape`, `dtypes`, `isnull()`, `sum()`

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('kc_house_data.csv')
df.head(10)
```

### Cell 2: Summary Statistics & Indexing (`loc` and `iloc`)
**Requirements:**
1. Print summary statistics for all numerical columns.
2. Use label-based indexing to select rows `0` through `99` and the columns `['price', 'bedrooms', 'bathrooms', 'sqft_living', 'yr_built']`.
3. Use position-based indexing to select rows `50` to `60` (exclusive) and the first `5` columns.
4. Find and print the number of unique zipcodes in the dataset.

**Tools & Functions to Use:** `describe()`, `loc[]`, `iloc[]`, `nunique()`

```python
display(df.describe())

loc_subset = df.loc[0:99, ['price', 'bedrooms', 'bathrooms', 'sqft_living', 'yr_built']]
display(loc_subset.head())
iloc_subset = df.iloc[50:60, 0:5]
display(iloc_subset)
print("Number of Unique Zipcodes:", df['zipcode'].nunique())
```

### Cell 3: Multi-Condition Filtering & Membership Testing
**Requirements:**
1. Filter houses built after `2000` (`yr_built > 2000`) with at least `4` bedrooms (`bedrooms >= 4`) and a `price` under `600000`.
2. Filter houses that either have a waterfront view (`waterfront == 1`) OR a building grade of `10` or higher (`grade >= 10`).
3. Filter houses located in any of these three zipcodes: `98004`, `98102`, or `98112`.
4. Print the number of rows in each filtered DataFrame.

**Tools & Functions to Use:** Boolean operators (`&`, `|`), `isin()`, `len()`

```python
filter_1 = df[(df['yr_built'] > 2000) & (df['bedrooms'] >= 4) & (df['price'] < 600000)]
filter_2 = df[(df['waterfront'] == 1) | (df['grade'] >= 10)]
filter_3 = df[df['zipcode'].isin([98004, 98102, 98112])]

print("Filter 1 Rows:", len(filter_1))
print("Filter 2 Rows:", len(filter_2))
print("Filter 3 Rows:", len(filter_3))
```

### Cell 4: Multi-Column Sorting & Frequency Counts
**Requirements:**
1. Sort the dataset by `grade` in descending order and then by `price` in ascending order, keeping the top 10 rows.
2. Count how many houses belong to each `grade` category.
3. Find the top 5 most frequent zipcodes in the dataset.

**Tools & Functions to Use:** `sort_values()`, `value_counts()`, `head()`

```python
top_10_sorted = df.sort_values(by=['grade', 'price'], ascending=[False, True]).head(10)
display(top_10_sorted)

print("Grade Counts:\n", df['grade'].value_counts())

print("Top 5 Zipcodes:\n", df['zipcode'].value_counts().head(5))
```

### Cell 5: Grouping & Multi-Metric Aggregation
**Requirements:**
1. Group the data by `condition` and compute the `mean`, `median`, `min`, and `max` of `price`.
2. Group the data by `waterfront` and compute the average `price` and average `sqft_living` simultaneously.
3. Group the data by `floors` and count the total number of houses (`id`) in each group.

**Tools & Functions to Use:** `groupby()`, `agg()`, `mean()`, `count()`

```python
print(df.groupby('condition')['price'].agg(['mean', 'median', 'min', 'max']))

print(df.groupby('waterfront')[['price', 'sqft_living']].mean())

print(df.groupby('floors')['id'].count())
```

## Section 2: NumPy (Cells 6 - 8)

### Cell 6: Array Creation, Descriptive Stats & Percentiles
**Requirements:**
1. Convert the `price` and `sqft_living` columns into NumPy arrays.
2. Calculate the mean, median, standard deviation, and variance of the `price` array.
3. Compute the 25th and 75th percentiles of the `price` array.
4. Count how many houses have a `sqft_living` strictly greater than the mean `sqft_living`.

**Tools & Functions to Use:** `np.array()`, `np.mean()`, `np.median()`, `np.std()`, `np.var()`, `np.percentile()`, `np.sum()`

```python
prices = np.array(df['price'])
living_area = np.array(df['sqft_living'])

print("Mean:", np.mean(prices))
print("Median:", np.median(prices))
print("Std:", np.std(prices))
print("Var:", np.var(prices))

print("25th and 75th Percentiles:", np.percentile(prices, [25, 75]))

print("Houses > Mean Living Area:", np.sum(living_area > np.mean(living_area)))
```

### Cell 7: Vectorized Operations & Mathematical Functions
**Requirements:**
1. Create a new column `price_per_sqft` by dividing `price` by `sqft_living`, rounded to `2` decimal places.
2. Create a column `total_rooms` as the sum of `bedrooms` and `bathrooms`.
3. Create a column `house_age` by subtracting `yr_built` from `2026`.
4. Create a column `log_price` by applying the base-10 logarithm to `price`.

**Tools & Functions to Use:** `np.round()`, `np.log10()`, vectorized arithmetic (`+`, `-`, `/`)

```python
df['price_per_sqft'] = np.round(df['price'] / df['sqft_living'], 2)
df['total_rooms'] = df['bedrooms'] + df['bathrooms']
df['house_age'] = 2026 - df['yr_built']
df['log_price'] = np.log10(df['price'])

display(df[['price_per_sqft', 'total_rooms', 'house_age', 'log_price']].head())
```

### Cell 8: Conditional Classification & Correlation
**Requirements:**
1. Create a column `is_renovated` set to `'Yes'` if `yr_renovated > 0`, otherwise `'No'`.
2. Use nested `np.where` to create a column `price_tier` classified as:
   - `'Budget'` if `price < 350000`
   - `'Mid-Range'` if `350000 <= price <= 750000`
   - `'Luxury'` if `price > 750000`
3. Compute and print the correlation matrix between `prices` and `living_area`.

**Tools & Functions to Use:** `np.where()`, `np.corrcoef()`, `value_counts()`

```python
df['is_renovated'] = np.where(df['yr_renovated'] > 0, 'Yes', 'No')

df['price_tier'] = np.where(
    df['price'] < 350000, 'Budget',
    np.where(df['price'] <= 750000, 'Mid-Range', 'Luxury')
)

print(df['is_renovated'].value_counts())
print(df['price_tier'].value_counts())
print(np.corrcoef(prices, living_area))
```

## Section 3: Matplotlib (Cells 9 - 10)

### Cell 9: Distribution & Relationship Plots (Histogram, Boxplot, Scatter)
**Requirements:**
1. Plot a histogram of `sqft_living` with `40` bins and add a vertical dashed red line representing the mean living area.
2. Plot a boxplot of `bedrooms` to inspect outliers.
3. Plot a scatter chart of `sqft_living` (x-axis) vs `price` (y-axis) with `alpha=0.4`, a title, axis labels, and a grid.

**Tools & Functions to Use:** `plt.figure()`, `plt.hist()`, `plt.axvline()`, `plt.boxplot()`, `plt.scatter()`, `plt.title()`, `plt.xlabel()`, `plt.ylabel()`, `plt.legend()`, `plt.grid()`, `plt.show()`

```python
plt.figure(figsize=(8, 5))
plt.hist(df['sqft_living'], bins=40, edgecolor='black')
plt.axvline(df['sqft_living'].mean(), color='red', linestyle='--', label='Mean Living Area')
plt.title('Distribution of Living Area')
plt.xlabel('sqft_living')
plt.ylabel('Frequency')
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(6, 5))
plt.boxplot(df['bedrooms'])
plt.title('Boxplot of Bedrooms')
plt.ylabel('Bedrooms')
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 5))
plt.scatter(df['sqft_living'], df['price'], alpha=0.4)
plt.title('Living Area vs Price')
plt.xlabel('sqft_living')
plt.ylabel('price')
plt.grid(True)
plt.show()
```

### Cell 10: Categorical & Trend Visualizations (Bar, Line, Pie)
**Requirements:**
1. Group by `grade` to calculate the mean `price`, then plot a bar chart of the results.
2. Filter houses built from `1950` onwards (`yr_built >= 1950`), group by `yr_built` to find the mean `price`, and plot a line chart showing the price trend over construction years.
3. Count the values in `price_tier` and plot a pie chart showing the percentage breakdown (`autopct='%1.1f%%'`).

**Tools & Functions to Use:** `plt.figure()`, `plt.bar()`, `plt.plot()`, `plt.pie()`, `plt.title()`, `plt.xlabel()`, `plt.ylabel()`, `plt.grid()`, `plt.show()`

```python
grade_mean = df.groupby('grade')['price'].mean()
plt.figure(figsize=(8, 5))
plt.bar(grade_mean.index, grade_mean.values)
plt.title('Mean Price by Grade')
plt.xlabel('Grade')
plt.ylabel('Mean Price')
plt.grid(True)
plt.show()

trend_1950 = df[df['yr_built'] >= 1950].groupby('yr_built')['price'].mean()
plt.figure(figsize=(9, 5))
plt.plot(trend_1950.index, trend_1950.values)
plt.title('Mean Price Trend (1950 Onwards)')
plt.xlabel('Year Built')
plt.ylabel('Mean Price')
plt.grid(True)
plt.show()

tier_counts = df['price_tier'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(tier_counts.values, labels=tier_counts.index, autopct='%1.1f%%')
plt.title('Price Tier Breakdown')
plt.show()
```
